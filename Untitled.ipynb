```python
# 1. Mediana original (pandas ignora los nulos al calcularla)
mediana_antes = df['horsepower'].median()
print(mediana_antes)

# 2. Valor más frecuente (la moda)
moda = df['horsepower'].mode()[0]
print(moda)

# 3. Rellenar los nulos con la moda
df['horsepower'] = df['horsepower'].fillna(moda)

# 4. Mediana de nuevo
mediana_despues = df['horsepower'].median()
print(mediana_despues)
```

```python
import numpy as np

# 1. Autos de Asia
asia = df[df['origin'] == 'Asia']

# 2 y 3. Dos columnas, primeras 7 filas
# 4. Array de NumPy
X = asia[['vehicle_weight', 'model_year']].head(7).to_numpy()

# 5. X transpuesta por X
XTX = X.T @ X

# 6. Inversa
XTX_inv = np.linalg.inv(XTX)

# 7. Vector y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. w = inv(XTX) · X^T · y
w = XTX_inv @ X.T @ y

# 9. Suma de los elementos
print(w)
print(w.sum())
```
